```python
#You are given a list of integer elements. Make a new list that will store a square of elements of the previous list. (With and without list comprehension) 
Input_list = [2,5,6,12] 
Output_list = [4,25,36,144]

#Without list comprehension
Input_list = [2,5,6,12] 
Output_list = []

for num in Input_list:
    square = num * num
    Output_list.append(square)
print(Output_list)


#With list comprehension
Output_list = [num *num for num in Input_list]
print(Output_list)
```

```python

# Write a Python program to check whether a list contains a sublist

list1 = [20, 30, 40, 50, 60, 70]
sublist1 = [30, 40]

found = False

for i in range(len(list1) - len(sublist1) + 1):
    if list1[i:i + len(sublist1)] == sublist1:
        found = True
        break

if found:
    print("Sublist found")
else:
    print("Sublist not found",sublist1)

```

What is the difference between shallow copy and deep copy? 
Shallow copy: Creates a new object but shares references to nested objects.
Deep copy: Creates a new object and recursively copies nested objects, so changes to nested objects don't affect the original.

```python
# Given a list of integers, create a new list where each element is replaced by the next largest  number in the original list. If there is no larger number, replace the element with -1. 
input_list = [4, 2, 6, 8, 1, 2] 
output_list = []

for i in range(len(input_list)):

    next_greater = -1

    for j in range(i + 1, len(input_list)):

        if input_list[j] > input_list[i]:
            next_greater = input_list[j]
            break

    output_list.append(next_greater)

print(output_list)

```

```python
#Write a function to reverse the elements within each tuple in a list of tuples. 
tuples = [(1, 2), (3, 4), (5, 6)] 
#Output: [(2, 1), (4, 3), (6, 5)] 

result= []

for item in tuples:
    result.append(item[::-1])

print(result)
```
